# Oblate single-bubble density slice

This chapter computes one diagnostic: the spatial integral of Eulerian energy density for a prescribed, constant-speed, unit-lapse shift profile with a quintic transition wall. It does **not** source the two-bubble CTC metric.

For axial semiaxis (a), speed (v), and dimensionless wall parameter (delta), the exact integral derived in the candidate note is

$$
\int \epsilon\,dV=-\frac{c^2 v^2a}{12G\delta}
\left(\frac{10}{7}+\frac{5\delta^2}{154}\right).
$$

It is independent of radial semiaxis (b) for this profile. That property is profile-specific, not a universal shape-optimization result. The output is not ADM mass, a total stress-energy audit, or an engineering energy budget.

```python
from pathlib import Path
import sys
import numpy as np
import plotly.graph_objects as go
from IPython.display import HTML, display

cwd = Path.cwd()
repo_root = next(parent for parent in (cwd, *cwd.parents) if (parent / "scripts/estimate_saucer_bubble_source.py").exists())
sys.path.insert(0, str(repo_root / "scripts"))
from estimate_saucer_bubble_source import calculate

base = calculate(a_m=10.0, b_m=20.0, wall_parameter=0.1, speed_fraction_c=0.1)
for key in ("volume_integral_eulerian_density_J", "volume_integral_equivalent_mass_kg", "local_density_order_scale_J_m3"):
    print(f"{key}: {base[key]:.6e}")
```

## How the diagnostic changes with wall thickness

```python
wall_parameters = np.geomspace(0.02, 0.5, 160)
energies = np.array([calculate(10.0, 20.0, d, 0.1)["volume_integral_eulerian_density_J"] for d in wall_parameters])
fig = go.Figure(go.Scatter(x=wall_parameters, y=-energies, mode="lines", name="magnitude"))
fig.update_layout(title="Integrated Eulerian-density magnitude vs wall parameter", xaxis_title="Dimensionless wall parameter δ", yaxis_title="|∫ ε dV| (joules)", xaxis_type="log", yaxis_type="log", template="plotly_white")
display(HTML(fig.to_html(full_html=False, include_plotlyjs="cdn")))
```

## Geometry-profile dependence

```python
axes = np.linspace(2, 30, 120)
fig = go.Figure()
for b in (8.0, 20.0, 40.0):
    vals = [-calculate(float(a), b, 0.1, 0.1)["volume_integral_eulerian_density_J"] for a in axes]
    fig.add_trace(go.Scatter(x=axes, y=vals, mode="lines", name=f"radial semiaxis b={b:g} m"))
fig.update_layout(title="Profile-specific dependence on saucer axes", xaxis_title="Axial semiaxis a (m)", yaxis_title="|∫ ε dV| (joules)", template="plotly_white")
display(HTML(fig.to_html(full_html=False, include_plotlyjs="cdn")))
```

The code imports the auditable formula from `scripts/estimate_saucer_bubble_source.py` in the repository source tree. Changing the metric class, lapse, shift profile, acceleration, or field support requires a fresh Einstein-tensor calculation; these cells do not validate those changes.
