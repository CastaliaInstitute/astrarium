# Two-bubble CTC: computable kinematics

This chapter reproduces the symmetric closure calculation discussed by Shoshany and Snodgrass and extends it algebraically to a nonzero spatial endpoint. The first model is the paper's symmetric benchmark. The second is a flat 1+1 kinematic extension only; it is **not** the paper's compact-support metric retargeted to Phoenix.

Let (U=u/c) be the relative inertial-frame speed and (V>1) the common average bubble separation speed in each leg's own asymptotic frame. The symmetric closed-loop finish time satisfies

$$
\frac{T_{\mathrm{finish}}}{T_2}
=\frac{2V-U(1+V^2)}{V-U},
\qquad
U>U_{\mathrm{crit}}=\frac{2V}{1+V^2}
$$

for a return to the initial spatial point before departure. For their smooth-profile example, (V=U[1+\omega(\alpha-1)/4]). The plotted “speed” is the average separation rate of a warp-supported region; the passenger worldline remains timelike.

For a target endpoint (C=(t_C,z_C)), the flat 1+1 relation is

$$
t_C=A T+B z_C,\quad
A=\frac{2V-U(1+V^2)}{V-U},\quad
B=\frac{VU-1}{V-U}.
$$

The Phoenix-scale inputs below use a constant 230 km/s solar-motion approximation over 29.6 years. This is not an ephemeris-grade Earth/witness state. The Phoenix event must lie on the full closed worldline, which this endpoint formula does not establish.

```python
import numpy as np
import plotly.graph_objects as go
from IPython.display import HTML, display

# Parameters of the paper's explicit smooth-profile example.
U0, alpha0, omega0 = 0.75, 10.0, 0.95
lookback_years = 29.6
solar_speed_km_s = 230.0
seconds_per_year = 365.25 * 24 * 3600
km_per_lightyear = 299_792.458 * seconds_per_year
z_target_ly = solar_speed_km_s * seconds_per_year / km_per_lightyear * lookback_years

def symmetric_ctc(U, V):
    """Return T_finish/T2 and the critical relative-frame speed."""
    return (2 * V - U * (1 + V**2)) / (V - U), 2 * V / (1 + V**2)

def endpoint_coefficients(U, V):
    """Flat 1+1 endpoint coefficients, with c=1."""
    return (2 * V - U * (1 + V**2)) / (V - U), (V * U - 1) / (V - U)

V0 = U0 * (1 + omega0 * (alpha0 - 1) / 4)
A0, B0 = endpoint_coefficients(U0, V0)
T_first = (-lookback_years - B0 * z_target_ly) / A0
D_first = V0 * T_first
print(f"V = {V0:.6f} c")
print(f"Ucrit = {2*V0/(1+V0**2):.8f}; chosen U = {U0:.3f}")
print(f"Tfinish/T2 = {A0:.8f}")
print(f"Approximate target displacement = {z_target_ly:.5f} ly")
print(f"Flat endpoint estimate: T2 = {T_first:.3f} y, first-leg separation = {D_first:.3f} ly")
```

## Closed-loop threshold versus average bubble speed

```python
V_grid = np.linspace(1.001, 8.0, 500)
Ucrit_grid = 2 * V_grid / (1 + V_grid**2)
fig = go.Figure()
fig.add_trace(go.Scatter(x=V_grid, y=Ucrit_grid, name="minimum frame speed", mode="lines"))
fig.add_hline(y=U0, line_dash="dash", annotation_text=f"example U={U0}")
fig.add_vline(x=V0, line_dash="dot", annotation_text=f"example V={V0:.3f}")
fig.update_layout(title="Symmetric CTC threshold", xaxis_title="Average bubble separation speed V / c", yaxis_title="Required relative frame speed Ucrit / c", yaxis_range=[0,1], template="plotly_white")
display(HTML(fig.to_html(full_html=False, include_plotlyjs="cdn")))
```

## Lookback scale for the cited parameter set

```python
lookbacks = np.linspace(1, 100, 200)
T2_values = lookbacks / (-A0)
D_values = V0 * T2_values
fig = go.Figure()
fig.add_trace(go.Scatter(x=lookbacks, y=D_values, mode="lines", name="first-leg separation"))
fig.add_trace(go.Scatter(x=[lookback_years], y=[D_first], mode="markers+text", text=[f"{D_first:.1f} ly"], textposition="top right", name="Phoenix nominal lookback"))
fig.update_layout(title="Interstellar scale implied by symmetric closure", xaxis_title="Lookback to same spatial point (years)", yaxis_title="First-leg separation (light-years)", template="plotly_white")
display(HTML(fig.to_html(full_html=False, include_plotlyjs="cdn")))
```

## Retargeting gap

Changing (z_C) shifts the flat endpoint estimate through coefficient (B), but it does not move the finite support profiles, recompute the lapse, or solve the geodesic equations in the published spacetime. A physical A.R.G.O. mission still needs a full CTG whose worldline intersects the Earth/witness world-tube at the Phoenix event, closes at the departure event, and passes the finite-hull and field-support checks.
